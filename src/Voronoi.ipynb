```python
import numpy as np
from scipy.spatial import Delaunay, delaunay_plot_2d, Voronoi, voronoi_plot_2d
import matplotlib.pyplot as plt
```

```python
# points = [
#     (0, 0),
#     (5, 2),
#     (3, 6),
#     (8, 7)
# ]

points = np.random.rand(5,2)  # random points in 2-D
```

```python
vor = Voronoi(points, furthest_site=False)
tri = Delaunay(points)
```

```python
# vor.point_region  # Index of the Voronoi region for each input point.
```

```python
# vor.vertices    # Coordinates of the Voronoi vertices.
```

```python
# vor.regions     # Indices of the Voronoi vertices forming each Voronoi region. -1 indicates vertex outside the Voronoi diagram
```

```python
# vor.ridge_vertices  # Indices of the Voronoi vertices forming each Voronoi ridge.
```

```python
# vor.ridge_points    #Indices of the points between which each Voronoi ridge lies.
```

```python
delaunay_plot_2d(tri)
voronoi_plot_2d(vor, ax=plt.gca(), point_size=10)
plt.show()
```
