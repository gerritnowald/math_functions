```python
import numpy as np
import matplotlib.pyplot as plt
```

input

```python
P0 = [0,1]
P1 = [3,2]
P2 = [5,0]
P3 = [4,-2]

t = 0.4
```

functions

```python
def lerp(P0,P1,t):
    return (1-t)*np.array(P0) + t*np.array(P1)

def lerp_plot(P0,P1,A):
    plt.plot([P0[0],P1[0]],[P0[1],P1[1]],"+--")
    plt.plot(A[0],A[1],"o")
```

linear

```python
A = lerp(P0,P1,t)

lerp_plot(P0,P1,A)
```

quadratic

```python
A0 = lerp(P0,P1,t)
A1 = lerp(P1,P2,t)

B = lerp(A0,A1,t)

lerp_plot(P0,P1,A0)
lerp_plot(P1,P2,A1)
lerp_plot(A0,A1,B)
```

cubic

```python
A0 = lerp(P0,P1,t)
A1 = lerp(P1,P2,t)
A2 = lerp(P2,P3,t)

B0 = lerp(A0,A1,t)
B1 = lerp(A1,A2,t)

C = lerp(B0,B1,t)

lerp_plot(P0,P1,A0)
lerp_plot(P1,P2,A1)
lerp_plot(P2,P3,A2)
lerp_plot(A0,A1,B0)
lerp_plot(A1,A2,B1)
lerp_plot(B0,B1,C)
```
